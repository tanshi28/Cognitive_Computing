```python
#prime numbers using function
def Prime(n):
    if(n<=3):
        return "Prime"
    for i in range(2,n//2 + 1):
        if(n%i==0):
            return "Not Prime"
    return "Prime"
print(Prime(57))
```

```python
#WAP using function that add all odd numbers from 1 to n, n is given by the user.
def addOdd(n):
    sum=0
    for i in range(1,n+1):
      if(i%2!=0):
          sum=sum+i
    return sum
          
print(addOdd(8))
```

```python
#How to use list, add, delete and search in the list.
L = ["Pratham",  'Sharma', 3.14,  3 ]
print ("Original List: ", L)
print ("Number of elements in list: ", len(L))
print ("Original List: ", L)
i=0
while i < len(L):
	print (L[i])
	i+=1
print ("Original List: ", L)
for i in range(0, len(L)):
	print (L[i])
print ("Original List --> ", L)
for s in L:
	print (s)
```

```python
#How to use Dictionary, add, delete, search in Dictionary
CGPA={1:8.9, 2:5.6, 4:6.7, 7:9.1, 8:5.3}
print ("Dictionary      --> ", CGPA)
print ("Num of elements --> ", len(CGPA))
print ("CGPA of 1       --> ", CGPA[1])
print ("CGPA of 4       --> ", CGPA[4])
print ("CGPA of 7       --> ", CGPA[7])
for k in CGPA:
	print ("CGPA of ", k, " --> ", CGPA[k])
print ("Dictionary --> ", CGPA)
print ("Keys       --> ", list(CGPA.keys()))
print ("Values     --> ", list(CGPA.values()))
print ("Original Dictionary --> ", CGPA)
print ("Is Key 2 Present    --> ", 2 in CGPA)
print ("Is Key 9 Present    --> ", 9 in CGPA)
```

```python
#How to use Dictionary, add, delete, search in Dictionary
CGPA={1:8.9, 2:5.6, 4:6.7, 7:9.1, 8:5.3}
print ("Dictionary      --> ", CGPA)
print ("Num of elements --> ", len(CGPA))
print ("CGPA of 1       --> ", CGPA[1])
print ("CGPA of 4       --> ", CGPA[4])
print ("CGPA of 7       --> ", CGPA[7])
print ("Dictionary --> ", CGPA)
print ("Keys       --> ", list(CGPA.keys()))
print ("Values     --> ", list(CGPA.values()))
```

```python
#Updating, Adding and Deleting from Diction
CGPA={1:8.9,2:5.6,4:6.7,7:9.1,8:5.3}
print ("Original Dictionary --> ", CGPA)

CGPA[4] = 9.2
print ("After Updating (4)  --> ", CGPA)

CGPA[3] = 8.6
print ("After Adding (3)    --> ", CGPA)

del CGPA[1]
print ("After Deleting (1)  --> ", CGPA)

CGPA.clear()
print ("After Clear         --> ", CGPA)

del CGPA
print ("After Delete        --> ", CGPA)
```
