```python
#Q1
r="1024170217"
L=[]
for i in range(0, len(r)):
    L.append(10*int(r[i]))
print(L)
```

```python
#using append
L.append(4)
print(L)

#using insert
L.insert(0,5)
print(L)
```

```python
#using pop
L.pop()
print(L)

#using remove
L.remove(5)
print(L)
```

```python
L.sort()
print(L)
L.sort(reverse=True)
print(L)
```

```python
print(L[-3:])
print(L[:3])
```

```python
avg=sum(L)/len(L)
NL=[]
for i in range(0,len(L)):
    if(L[i]>avg):
        NL.append(L[i])
print(NL)
```

```python
#Q4
r="10241702"
A=set()
B=set()
for i in r:
    A.add(7*int(i))
    B.add(9*int(i))
print(A)
print(B)
```

```python
A.union(B)
```

```python
A.intersection(B)
```

```python
print(A.difference(B))
print(B.difference(A))
```

```python
print(A.symmetric_difference(B))
print(B.symmetric_difference(A))
```

```python
A.issubset(B)
```

```python
B.issuperset(A)
```
