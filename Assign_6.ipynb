```python
# Q1
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
corrected = temperature + 2

print("Original temperatures:", temperature)
print("Corrected temperatures:", corrected)
fahrenheit = (9 / 5) * corrected + 32
print("Temperatures in Fahrenheit:", fahrenheit)
greater_than_32 = corrected[corrected > 32]
print("Readings greater than 32°C:", greater_than_32)
count = np.sum(corrected> 32)
print("Number of readings greater than 32°C:", count)
```

```python
# Q2
steps = np.array([ [5000, 6200, 7100], [8000, 7500, 9000], [4500, 5100, 4800], [9000, 8500, 9500]])
total_steps=np.sum(steps)
print("Total steps ",total_steps)
average_steps=np.mean(steps)
print("Average steps ",average_steps)
max_steps=np.max(steps)
print("Maximum steps ",max_steps)
min_steps=np.min(steps)
print("Minimum steps ",min_steps)
daily_total = np.sum(steps, axis=0)
print("Total steps for each day:", daily_total)
user_total = np.sum(steps, axis=1)
print("Total steps for each user:", user_total)

max_index = np.argmax(steps)
print("Index of maximum value:", max_index)
position = np.unravel_index(np.argmax(steps), steps.shape)
print("User/day position:", position)
print("Maximum steps:", steps[position])
```

```python
# Q3
original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

# b) Slice
subset = original[1:4]
print("Subset:", subset)

# c) Modify view
subset[0] = 999

print("Original:", original)
print("Subset:", subset)

# d) Copy
copied_array = original[1:4].copy()
copied_array[0] = 500

print("Original:", original)
print("Copied array:", copied_array)

# e) arange and reshape
matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)

# f) Indexing and slicing
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

# g) flatten and ravel
flattened = matrix.flatten()
raveled = matrix.ravel()

print("Flatten:", flattened)
print("Ravel:", raveled)

# h) Modify ravel
raveled[0] = 1000

print("Ravel:", raveled)
print("Original matrix:")
print(matrix)

# i) Modify flatten
flattened[1] = 500

print("Flatten:", flattened)
print("Original matrix:")
print(matrix)

# j) Array properties
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
# Question 4
y = np.array([40, 65, 30, 85])

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

# a)
print("Shape:", X.shape)
print("Dimensions:", X.ndim)

# b)
print("X.T:",X.T)

# c)
print("X.T @ X:",X.T @ X)

# d)
inverse = np.linalg.inv(X.T @ X)
print("Inverse:",inverse)

# e)
beta = inverse @ X.T @ y
print("Coefficients:", beta)

# f)
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])

# g)
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
