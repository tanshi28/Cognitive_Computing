```python
# Question 1
import numpy as np
arr=np.array([1,2,3,4,5])
print(arr)
#Addition of 2
addn=arr+2
print(addn)
# Multiplication by 3
multn= arr*3
print(multn)
#Division by 2
divn = arr/2
print(divn)
```

```python
# Question 2 (i)
x = np.array([1,2,3,4,5,1,2,1,1,1])
values,counts=np.unique(x,return_counts=True)
most_freq=values[np.argmax(counts)]
ind=np.where(x==most_freq)[0]
print("Most frequent value:", most_freq)
print("Frequency:", np.max(counts))
print("Indices:", ind)
```

```python
# Question 2 (ii)
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values,counts=np.unique(y,return_counts=True)
most_freq=values[np.argmax(counts)]
ind=np.where(y==most_freq)[0]
print("Most frequent value:", most_freq)
print("Frequency:", np.max(counts))
print("Indices:", ind)
```

```python
# Question 4
ucs420_Tanshi=np.linspace(10,100,25)
print("array:",ucs420_Tanshi)
print("Dimensions:", ucs420_Tanshi.ndim)
print("Shape:", ucs420_Tanshi.shape)
print("Total elements:", ucs420_Tanshi.size)
print("Data type:", ucs420_Tanshi.dtype)
print("Total bytes:", ucs420_Tanshi.nbytes)
# Transpose using reshape
transpose_reshape = ucs420_Tanshi.reshape(1, 25).T
print("Transpose using reshape:",transpose_reshape)

# Transpose using T
transpose_T = ucs420_Tanshi.T
print("Using T attribute:",transpose_T)
```

```python
# Question 5
import numpy as np

ucs420_Tanshi = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:",ucs420_Tanshi)

print("Mean:", np.mean(ucs420_Tanshi))
print("Median:", np.median(ucs420_Tanshi))
print("Maximum:", np.max(ucs420_Tanshi))
print("Minimum:", np.min(ucs420_Tanshi))
print("Unique elements:", np.unique(ucs420_Tanshi))

# Reshape to 4 rows and 3 columns
reshaped_ucs420_Tanshi = ucs420_Tanshi.reshape(4, 3)
print("Reshaped Array:",reshaped_ucs420_Tanshi)

# Resize to 2 rows and 3 columns
resized_ucs420_Tanshi = np.resize(ucs420_Tanshi, (2, 3))
print("Resized Array:",resized_ucs420_Tanshi)
```
